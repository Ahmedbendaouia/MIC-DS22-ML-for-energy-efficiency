```python
def value_by_poste(data):
    list_courant,list_debit_eau,list_solide,list_puissance,list_time,list_poste = [],[],[],[],[],[]
    time_start = pd.to_datetime("2022-01-28 00:00:00")
    columns = data.columns
    while(time_start <= pd.to_datetime("2022-10-20 17:01:00")):
        for i in range(3):
            interv = data[data[columns[0]].between(time_start,time_start + timedelta(hours=8))]
            if interv.shape[0] != 0:
                list_courant.append(interv["COURANT ValueY"].mean())
                list_debit_eau.append(interv["DEBIT EAU ValueY"].mean())
                list_solide.append(interv["SOLIDE ValueY"].mean())
                list_puissance.append(interv["PUISSANCE ValueY"].mean())
                list_time.append(time_start)
                #list_poste.append(f"Poste {i+1}")
            time_start = time_start + timedelta(hours=8)
    new_data = pd.DataFrame(data=zip(list_time,list_courant,list_debit_eau,list_solide,list_puissance),columns=["Time","COURANT ValueY","DEBIT EAU ValueY","SOLIDE ValueY","PUISSANCE ValueY"])
    return new_data
```

```python
new_data = value_by_poste(data)
new_data
```

```python
data_refus_tamis = pd.read_excel("De 11_02 à 17_02 2022\Refus Tamis\TB_D80_et_performances_Usine_Février_2022.xlsx")
data_refus_tamis_octobre = pd.read_excel("De 15_09 à 20_10 2022\Refus Tamis\TB_D80_et_performances_Usine_Octobre_2022.xlsx")
data_refus_tamis_septembre = pd.read_csv("De 15_09 à 20_10 2022\Refus Tamis\data_refus_tamis_septembre.csv")
data_refus_tamis_aout = pd.read_excel("De 23_08 à 15_09 2022\Refus Tamis\TB_D80_et_performances_Usine_Août_2022.xlsx")
data_refus_tamis_septembre_2 = pd.read_excel("De 23_08 à 15_09 2022\Refus Tamis\TB_D80_et_performances_Usine_Septembre_2022.xlsx")
data_refus_tamis_janvier = pd.read_excel("De 28_01 à 03_02 2022\Refus Tamis\TB_D80_et_performances_Usine_Janvier_2022.xlsx")
data_refus_tamis_fevrier = pd.read_excel("De 28_01 à 03_02 2022\Refus Tamis\TB_D80_et_performances_Usine_Février_2022.xlsx")
```

```python
data_refus_tamis_septembre['Jours'] = pd.to_datetime(data_refus_tamis_septembre['Jours'], format="%Y/%m/%d %H:%M:%S")
```

```python
def process_refu_tamis(data):
    data.dropna(how='all',inplace=True)
    del data["Unnamed: 0"]
    data.rename(columns={'Unnamed: 1':"Jours", 'Unnamed: 2':"Poste",'Unnamed: 3':"Refus Tamis +500 en % QA"}, inplace=True)
    data.drop(index=1,inplace=True)
    for i,line in enumerate(data.index):
        if data.iloc[i]["Jours"] is np.nan:
            data.iloc[i]["Jours"] = data.iloc[i-1]["Jours"]
        if (data.iloc[i]["Refus Tamis +500 en % QA"] is np.nan):
            data.iloc[i]["Refus Tamis +500 en % QA"] = data.iloc[i-1]["Refus Tamis +500 en % QA"]
    data.drop(data.tail(3).index,inplace = True)
    data['Jours'] = pd.to_datetime(data['Jours'], format="%Y/%m/%d %H:%M:%S")
    data["Refus Tamis +500 en % QA"] = data["Refus Tamis +500 en % QA"].astype(float)
```

```python
process_refu_tamis(data_refus_tamis)
process_refu_tamis(data_refus_tamis_octobre)
process_refu_tamis(data_refus_tamis_aout)
process_refu_tamis(data_refus_tamis_septembre_2)
process_refu_tamis(data_refus_tamis_janvier)
process_refu_tamis(data_refus_tamis_fevrier)
```

```python
def add_refus_tamis(refus_tamis,data):
    for i,line in enumerate(refus_tamis.index):
        v1 = refus_tamis.iloc[i]["Jours"]
        if i==0:
            v3 = v1
            for j in range(3):
                data.loc[data.Time == v3,'refus_tamis']= list(refus_tamis[refus_tamis["Jours"] == v1]["Refus Tamis +500 en % QA"])[j]
                v3 = v3 + timedelta(hours=8)
            v2 = v1
        else:
            if v1 != v2:
                v3 = v1
                for j in range(3):
                    data.loc[data.Time == v3,'refus_tamis']= list(refus_tamis[refus_tamis["Jours"] == v1]["Refus Tamis +500 en % QA"])[j]
                    v3 = v3 + timedelta(hours=8)
                v2 = v1  
        
```
