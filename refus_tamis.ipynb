```python
mu=0.0
std = 0.005
def gaussian_noise(x,mu,std):
    noise = np.random.normal(mu, std)
    x_noisy = x + noise
    return x_noisy 
```

11_02 à 17_02

```python
data_refus_tamis = pd.read_excel("De 11_02 à 17_02 2022\Refus Tamis\TB_D80_et_performances_Usine_Février_2022.xlsx")
```

```python
def process_refu_tamis(data):
    data.dropna(how='all',inplace=True)
    del data["Unnamed: 0"]
    data.rename(columns={'Unnamed: 1':"Jours", 'Unnamed: 2':"Poste",'Unnamed: 3':"Refus Tamis +500 en % QA"}, inplace=True)
    data.drop(index=1,inplace=True)
    for i,line in enumerate(data.index):
        if data.iloc[i]["Jours"] is np.nan:
            data.iloc[i]["Jours"] = data.iloc[i-1]["Jours"]
        if (data.iloc[i]["Refus Tamis +500 en % QA"] is np.nan) and not ((pd.to_datetime(data.iloc[i]["Jours"]) >= (pd.to_datetime("2022-09-15 00:00:00"))) and (pd.to_datetime(data.iloc[i]["Jours"]) <= pd.to_datetime("2022-09-30 00:00:00"))):
            data.iloc[i]["Refus Tamis +500 en % QA"] = data.iloc[i-1]["Refus Tamis +500 en % QA"]
    data.drop(data.tail(3).index,inplace = True)
    data['Jours'] = pd.to_datetime(data['Jours'], format="%Y/%m/%d %H:%M:%S")
    data["Refus Tamis +500 en % QA"] = data["Refus Tamis +500 en % QA"].astype(float)
```

```python
process_refu_tamis(data_refus_tamis)
```
